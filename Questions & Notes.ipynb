# Possible Key Insights

### 1. Chicken-in-egg / codependency problem

People locate where companies are. And companies locate where people are.

Related to #2, companies with small/medium/somewhat large market capitalization/valuation may not want to locate to industry hubs (cost likely being primary motivator). (related question: of the clients to savills, what is their company size?). But it may be risky.

**So our "one key insight" could be to model (basically!) and industry, and determine if a company could/should move to different city, based on how it relates to other companies in same industry that have relocated.**

Can plan for variables like "good state laws" by pre-choosing. But, many things vary over time:

- Lack of talent
- Lack of other companies
- Lack of transport (road/train for employees, airports for execs)
- Lack of amnemities
- Safety of neighborhood / covid related thing I forgot

One way to attack this, is to search for the types of companies of an industry. Take finance, you have:

- Companies operating stock markets (NASDAQ, ICE (NYSE))
- Companies operating brokerage firms (Fidelity, Vanguard)
- Companies offering venues for dark pools
- Companies rating marketable securities (Moody's, etc.)
- Companies defined as a market maker

Say, given a relationship between two companies, A and B. If A depends on B's services, and B relocates to different city, it is now more OK for A to relocate to same place.

Potential problems:

- Industries like Finance often operate online (Excluding time-based arbitrage, HFT, etc.)
- Trend of bigger&bigger companies, that offer more varying services "all-in-one platform" (Vanguard is a brokerage firm for individuals, holds many pension funds for cities, etc., day-trading platform... _kinda_)
- Another reason, usually these companies don't organically increase in this cause-effect way, everything usually grows in tandem, and growth is more "personal" and can't really generalize (depeds on personal relationships?)
- Sucks, label & analyze each individual company

These make the relationships more complex and difficult to model. Any simplification would very likely be on verge of being incorrect. Doesn't use proprietary data as much
